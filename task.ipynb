```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# TASK 1: EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# Load any public dataset (CSV format).
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
print("Dataset loaded successfully!")

# Display the number of rows and columns
# df.shape 
print("Dataset shape:")
print(df.shape)

# df.head() displays the first 5 rows of the dataset.
print("displaying first five rows:")
print(df.head())

# 3. Display information about the dataset
print("Dataset Information:")
df.info()

# Descriptive Statistics
# Select only numerical columns.
numeric_columns = df.select_dtypes(include=np.number).columns


# Calculate the mean (average) of each numerical column.
print("Mean:")
print(df[numeric_columns].mean())

# Calculate the median of each numerical column.
print("Median:")
print(df[numeric_columns].median())

# Calculate the standard deviation.
print("Standard Deviation:")
print(df[numeric_columns].std())

# Display a complete statistical summary.
print("Complete Descriptive Statistics:")
print(df.describe())

# 5. Check for missing values
# Count the number of missing values in every column.
print("Missing Values:")
print(df.isnull().sum())

# Calculate the percentage of missing values.
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Values Percentage:")
print(missing_percentage)

# Handle missing values

# Replace missing Age values with the median Age.
df["Age"] = df["Age"].fillna(df["Age"].median())

# Replace missing Embarked values with the most frequent value.
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Replace missing Fare values with the median Fare.
df["Fare"] = df["Fare"].fillna(df["Fare"].median())

# Check missing values again after cleaning.
print("Missing Values After Cleaning:")
print(df.isnull().sum())

# Count duplicate rows.
duplicates = df.duplicated().sum()

print("Number of duplicate rows:")
print(duplicates)

# Remove duplicate rows.
df = df.drop_duplicates()

print("Duplicate rows removed successfully.")

# Detect outliers using the IQR method
# Create a function to detect outliers.
def detect_outliers(data, column):
    # Calculate the first quartile.
    Q1 = data[column].quantile(0.25)
    # Calculate the third quartile.
    Q3 = data[column].quantile(0.75)
    # Calculate the Interquartile Range.
    IQR = Q3 - Q1
    # Calculate the lower boundary.
    lower_limit = Q1 - 1.5 * IQR
    # Calculate the upper boundary.
    upper_limit = Q3 + 1.5 * IQR

    # Select values below the lower limit
    # or above the upper limit.
    outliers = data[
        (data[column] < lower_limit) |
        (data[column] > upper_limit)
    ]

    return outliers, lower_limit, upper_limit


# Detect Age outliers
age_outliers, age_lower, age_upper = detect_outliers(df, "Age")

print("Age Outlier Analysis:")
print("Number of Age outliers:", len(age_outliers))
print("Age lower limit:", age_lower)
print("Age upper limit:", age_upper)

# Detect Fare outliers

fare_outliers, fare_lower, fare_upper = detect_outliers(df, "Fare")

print("Fare Outlier Analysis:")
print("Number of Fare outliers:", len(fare_outliers))
print("Fare lower limit:", fare_lower)
print("Fare upper limit:", fare_upper)

# Visualize Age distribution
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Age"],
    bins=30,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

# Display the graph.
plt.show()

# Visualize Fare distribution

plt.figure(figsize=(8, 5))

# Create a histogram showing the distribution of Fare.
sns.histplot(
    df["Fare"],
    bins=30,
    kde=True
)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()

# Visualize Age outliers using a boxplot
plt.figure(figsize=(8, 5))

# Create a boxplot for Age.
sns.boxplot(x=df["Age"])
plt.title("Age Outliers")
plt.xlabel("Age")

plt.show()

#  Visualize Fare outliers using a boxplot
plt.figure(figsize=(8, 5))
# Create a boxplot for Fare.
sns.boxplot(x=df["Fare"])
plt.title("Fare Outliers")
plt.xlabel("Fare")
plt.show()

# Survival distribution
plt.figure(figsize=(7, 5))
# Create a count plot for the Survived column.
sns.countplot(
    data=df,
    x="Survived"
)

plt.title("Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()

# Correlation heatmap
correlation = df.select_dtypes(include=np.number).corr()
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.show()
```
